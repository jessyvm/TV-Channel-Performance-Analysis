```python
import pandas as pd

df = pd.read_csv("Food_Prices.csv", encoding="latin1")
print(df)
```

```python
print(df.head())
print(df.tail())
```

```python
df['adm1_name']=df['adm1_name'].fillna('Unknown')
print(df['adm1_name'])
```

```python
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
print("\nStatistical Summary:")
print(df.describe())    
```

```python
print("\nAverage prize:")
print(df["mp_price"].mean())
print("\nHighest prize:")
print(df["mp_price"].max())
print("\nLowest price:")
print(df["mp_price"].min())
```

```python
print("\nCustomer IDs:")
print(df["cur_id"].unique())
```

```python
print("\nCustomer Counts:")
print(df["cur_id"].value_counts())
```

```python
import matplotlib.pyplot as plt
df["mp_year"].value_counts().plot(kind="bar", color="green",edgecolor="black", alpha=0.7, width=0.5,)

plt.title("Food Prices by Year")
plt.xlabel("mp_year")
plt.ylabel("mp_price")
plt.show()
```
