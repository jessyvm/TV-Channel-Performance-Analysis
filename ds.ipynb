```python
import matplotlib.pyplot as plt
songs = ["APT.", "Espresso", "Die With A Smile", "Houdini", "Birds Of A Feather"]
streams = [245, 195, 225, 100, 185]
plt.bar(songs, streams, color='green', edgecolor='black', alpha=0.7, width=0.5, linestyle='--', linewidth=1.5, hatch='//', align='center', orientation='vertical', zorder=3)
plt.title("Trending Songs")
plt.ylabel("Streams (Millions)")
plt.xticks(rotation=30)
plt.show()
```

```python
import pandas as pd
df=pd.read_csv("Ecommerce dataset.csv")
print(df)
```

```python
print(df.info)
print(df.shape)
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates()
```

```python
df.duplicated()
```

```python
df['delivery_days']=df['delivery_days'].fillna(df['delivery_days'].mean())
print(df['delivery_days'])
df['estimated_delivery_days']=df['estimated_delivery_days'].fillna(df['estimated_delivery_days'].mean())
print(df['estimated_delivery_days'])
```

```python
print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
print("\nStatistical Summary:")
print(df.describe())    
```

```python
df["customer_age"] = pd.to_numeric(df["customer_age"], errors="coerce")
df = df[df["customer_age"] >= 18]
print(df)
```

```python
print(df.head())
print(df.tail())
```

```python
print("\nAverage Age:")
print(df["customer_age"].mean())
print("\nAverage sales:")
print(df["gross_sales"].mean())
print("\nHighest sales:")
print(df["loyalty_points_earned"].max())
print("\nLowest sales:")
print(df["loyalty_points_earned"].min())
```

```python
print("\ncustomer_type:")
print(df["customer_type"].unique())
```

```python
print("\nCustomer Age Counts:")
print(df["customer_age"].value_counts(24))
```

```python
df["customer_order_count"].value_counts().plot(kind="bar", color="green", edgecolor="black", alpha=0.7, width=0.5, linestyle="--", linewidth=1.5, hatch="//", align="center", zorder=3)
plt.title("Customer Order Count Distribution")
plt.xlabel("Order Count")
plt.ylabel("Customer Count")
plt.show()
```
